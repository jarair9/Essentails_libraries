```python
# Install Pandas using commands `pip install pandas` 

import pandas
# But instead of importing pandas and using whole word as class developer use pd short for pandas.
import pandas as pd

data = {
    "name": ["jarair","haseeb","hamza", "farhan","ahmad"],
    "Age": [17,18,29,10,27],
    "class" : [11,12,10,8,9]
    
}

# Important: Series is 1D data so i use one column
df = pd.Series(data["name"], index=["class1","class2","class3","class4","class5"])


# .loc is label based filtering
df.loc["class3"]

# .iloc index based
df.iloc[0]


# Codition adding
df = pd.Series(data["Age"])
df[df > 17]
```

## Creating and Manipulating DataFrames (Two-Dimensional Data)

A **DataFrame** is a two-dimensional labeled grid, table, or tabular data structure with rows and columns.

```python
data = {
    "name": ["jarair","haseeb","hamza", "farhan","ahmad"],
    "Age": [17,18,29,10,27],
    "class" : [11,12,10,8,9]
    
}

# Important: Dataframe is 2D Data
df = pd.DataFrame(data, index=["class1","class2","class3","class4","class5"])


# .loc is label based filtering
df.loc["class3"]

# .iloc index based
df.iloc[0]

# Adding new Column
df["Marks"] = [400,355,677,909,1021]

# Adding Row
new_member = pd.DataFrame({
    "name": ["mohsin"], 
    "Age":80, 
    "class":34, 
    "Marks": [100]
}, index=["class6"])

df = pd.concat([df,new_member])
df
```

```python
import pandas as pd


# 2. Display the original Data
print("=== ORIGINAL DATAFRAME ===")
display(df)
print("\n" + "="*50 + "\n")

# Mean/Average
print("--- 1. Mean (Average) ---")
display(df.mean(numeric_only=True))
print("\n")

# Sum
print("--- 2. Sum ---")
display(df.sum(numeric_only=True))
print("\n")

# Minimum
print("--- 3. Minimum ---")
display(df.min())
print("\n")

# Maximum
print("--- 4. Maximum ---")
display(df.max())
print("\n")

# Count
print("--- 5. Count ---")
display(df.count())
print("\n")

# Column Aggregation
print("--- 6. Column Aggregation ---")
avg_marks = df['Marks'].mean()
print(f"The average of Marks column is: {avg_marks}")
print("\n" + "="*50 + "\n")

# Groupby Function
print("--- 7. Groupby (Average Age and Marks per Class) ---")
display(df.groupby('class')[['Age', 'Marks']].mean())
```
